# 绘制 GEA 超参数折线图脚本

import matplotlib.pyplot as plt
import numpy as np
def plot_line():

    # label在图示(legend)中显示。若为数学公式,则最好在字符串前后添加"$"符号
    # color：b:blue、g:green、r:red、c:cyan、m:magenta、y:yellow、k:black、w:white、、、
    # 线型：-  --   -.  :    ,
    # marker：.  ,   o   v    <    *    +    1


    # 数据
    x = np.array([0.1, 0.3, 0.5, 0.7,0.9])
    # x = np.array([30, 45, 60,75,90])
    baseline = np.array([68.795, 68.795, 68.795, 68.795,68.795 ]) # baseline HOTA
    # Ours = np.array([68.993, 68.914, 68.836, 69.078, 68.749])   # p
    # Ours = np.array([68.867, 68.862, 68.909, 68.859, 69.006])   # RL
    # Ours = np.array([68.840, 68.862, 69.081, 69.002, 68.770])   # Sh
    Ours = np.array([68.539, 68.612, 68.865, 69.155, 68.852])   # thetah
    # Ours = np.array([ 68.887, 68.481, 68.955, 68.915 ])   # theta

    # 设置画布
    plt.figure(figsize=(5, 4))
    # linestyle = "-"
    plt.grid(linestyle="-.")  # 设置背景网格线为虚线
    ax = plt.gca()  # 移动坐标系
    # ax.spines['top'].set_visible(False)  # 去掉上边框
    # ax.spines['right'].set_visible(False)  # 去掉右边框


    # 绘制折线
    linewidth = 1.5 # 线宽
    markersize = 7  # 标记点大小
    plt.plot(x, baseline, marker='s', markersize=markersize, color="blue", label="baseline", linewidth=linewidth,linestyle='dashed')
    plt.plot(x, Ours, marker='X', markersize=markersize, color="tomato", label="Ours", linewidth=linewidth)


    # # 设置坐标轴
    plt.xlim(0.06, 0.94  )  # 设置x轴的范围
    # group_labels = ['0.1', '0.3', '0.5', '0.7', '0.9']
    group_labels = ['$30^\circ$', '$45^\circ$', '$60^\circ$', '$75^\circ$', '$90^\circ$']
    plt.xticks(x, group_labels, fontsize=15)  # 默认字体大小为10

    # plt.xlim(27, 93  )  # 设置x轴的范围
    # group_labels = [ '0.25', '0.5', '1.0','2.0']
    # plt.xticks(x, group_labels, fontsize=15)  # 默认字体大小为10

    plt.ylim(68.4, 69.2)
    y_ticks = [ 68.4, 68.6, 68.8,69.0,69.2]
    y_lables = ['68.4', '68.6', '68.8', '69.0', '69.2']
    plt.yticks(np.array(y_ticks), y_lables, fontsize=15)
    plt.ylabel(f"HOTA", fontsize=15)
    # plt.title("example", fontsize=12, fontweight='bold')  # 默认字体大小为12
    # plt.text(1, label_position, dataset,fontsize=25, fontweight='bold')
    plt.xlabel(r"$\theta_{h}$", fontsize=15)


    # 图例
    # 显示各曲线的图例 loc=3 lower left
    plt.legend(loc=0, numpoints=1, ncol=2)    # 设置图例的位置
    leg = plt.gca().get_legend()
    ltext = leg.get_texts()
    plt.setp(ltext, fontsize=15)
    # plt.setp(ltext, fontsize=25, fontweight='bold')  # 设置图例字体的大小和粗细
    plt.tight_layout()  # 自动调整标签分布


    # 保存
    plt.savefig(f'./line-theta.svg', format='svg')  # 建议保存为svg格式,再用inkscape转为矢量图emf后插入word中
    plt.show()

plot_line()
